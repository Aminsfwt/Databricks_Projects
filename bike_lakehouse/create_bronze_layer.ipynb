### Import the configration file

##### file that contains the path of the files and the target of the tables

```python
from config_file import INGESTION_CONFIG
```

%md
## Create Function to read files from the volume and load it into table

```python
for item in INGESTION_CONFIG:
    print(f"Start Ingesting {item['source']} to {item['table']}")
    
    # Read the files into dataframe
    df = spark.read \
        .option("header", True) \
        .option("inferSchema", True) \
        .csv(item['path']
    )

    # write it into a table
    df.write \
        .mode("overwrite") \
        .saveAsTable(f"bike_project_catalog.bronze.{item['table']}")
```
